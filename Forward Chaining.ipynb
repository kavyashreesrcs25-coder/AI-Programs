```python

def main():
    print("FORWARD CHAINING IN FIRST ORDER LOGIC")

    # Get initial facts
    n = int(input("\nEnter number of facts: "))
    KB = []

    for i in range(n):
        fact = input("Enter fact (Predicate, arg1, arg2): ")
        parts = [x.strip() for x in fact.split(",")]
        KB.append(tuple(parts))

    # Get rule: Parent(x,y) AND Parent(y,z) -> Grandparent(x,z)
    print("\nRule: Parent(x,y) AND Parent(y,z) -> Grandparent(x,z)")

    # Get query
    query = input("\nEnter query (Predicate, arg1, arg2): ")
    query = tuple(x.strip() for x in query.split(","))

    # Apply forward chaining
    new = []

    for fact1 in KB:
        for fact2 in KB:
            if (fact1[0] == "Parent" and
                fact2[0] == "Parent" and
                fact1[2] == fact2[1]):

                result = ("Grandparent", fact1[1], fact2[2])

                if result not in KB and result not in new:
                    new.append(result)

    KB.extend(new)

    print("\nNewly Inferred Facts:")
    for fact in new:
        print(fact)

    print("\nFinal Knowledge Base:")
    for fact in KB:
        print(fact)

    print("\nQuery:", query)

    if query in KB:
        print("Query Proved: True")
    else:
        print("Query Proved: False")


if __name__ == "__main__":
    main()
```
