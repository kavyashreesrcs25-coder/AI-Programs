```python

def parse_fact(expression):
    name, values = expression.strip().replace(")", "").split("(")
    arguments = tuple(x.strip() for x in values.split(","))
    return (name.strip(),) + arguments


def main():
    print("FORWARD CHAINING IN FIRST ORDER LOGIC")

    n = int(input("\nEnter number of facts: "))
    KB = set()

    for i in range(n):
        fact = input("Enter fact: ")
        KB.add(parse_fact(fact))

    query = parse_fact(input("\nEnter query: "))

    print("\nInitial Knowledge Base:")
    for fact in sorted(KB):
        print(fact)

    # Apply Forward Chaining rules repeatedly
    while True:
        new = set()

        # Rule 1: Missile(x) -> Weapon(x)
        for fact in KB:
            if fact[0] == "Missile":
                new.add(("Weapon", fact[1]))

        # Rule 2: Missile(x) AND Owns(A,x)
        #         -> Sells(Robert,x,A)
        for missile in KB:
            if missile[0] == "Missile":
                x = missile[1]

                for owns in KB:
                    if owns == ("Owns", "A", x):
                        new.add(("Sells", "Robert", x, "A"))

        # Rule 3: Enemy(x,America) -> Hostile(x)
        for fact in KB:
            if fact[0] == "Enemy" and fact[2] == "America":
                new.add(("Hostile", fact[1]))

        # Rule 4: American(p) AND Weapon(q)
        # AND Sells(p,q,r) AND Hostile(r) -> Criminal(p)
        for american in KB:
            if american[0] == "American":
                p = american[1]

                for weapon in KB:
                    if weapon[0] == "Weapon":
                        q = weapon[1]

                        for sells in KB:
                            if sells[0] == "Sells" and sells[1] == p and sells[2] == q:
                                r = sells[3]

                                if ("Hostile", r) in KB:
                                    new.add(("Criminal", p))

        # Remove facts already present
        new = new - KB

        if not new:
            break

        print("\nNewly Inferred Facts:")
        for fact in sorted(new):
            print(fact)

        KB.update(new)

    print("\nFinal Knowledge Base:")
    for fact in sorted(KB):
        print(fact)

    print("\nQuery:", query)

    if query in KB:
        print("Query Proved: TRUE")
    else:
        print("Query Proved: FALSE")


if __name__ == "__main__":
    main()
```
