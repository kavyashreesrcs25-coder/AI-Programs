```python
initial = (1,2,3,4,0,6,7,5,8)
goal = (1,2,3,4,5,6,7,8,0)

def moves(s):
    p = s.index(0)
    r, c = divmod(p, 3)
    ans = []

    for dr, dc, name in [(1,0,"Down"), (0,1,"Right"),
                          (-1,0,"Up"), (0,-1,"Left")]:
        nr, nc = r+dr, c+dc

        if 0 <= nr < 3 and 0 <= nc < 3:
            q = nr*3 + nc
            x = list(s)
            x[p], x[q] = x[q], x[p]
            ans.append((name, tuple(x)))

    return ans

def dfs(s, path, visited):
    if s == goal:
        return path

    visited.add(s)

    for move, new_s in moves(s):
        if new_s not in visited:
            result = dfs(new_s, path+[move], visited)
            if result:
                return result

    return None

answer = dfs(initial, [], set())

print("DFS Solution:", answer)
print("Number of moves:", len(answer))
```
