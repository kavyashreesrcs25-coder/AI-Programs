```python
from itertools import product

def implies(a, b):
    return not a or b


def main():
    print("P Q R | Q->P | P->~Q | Q|R | KB")
    print("-----------------------------------")

    models = []

    for P, Q, R in product([True, False], repeat=3):
        s1 = implies(Q, P)
        s2 = implies(P, not Q)
        s3 = Q or R

        KB = s1 and s2 and s3

        print(
            int(P), int(Q), int(R), "|",
            int(s1), "   |",
            int(s2), "    |",
            int(s3), "  |",
            int(KB)
        )

        if KB:
            models.append((P, Q, R))

    print("\nModels where KB is true:")
    for model in models:
        print(model)

    entails_R = all(R for P, Q, R in models)
    entails_R_implies_P = all(
        implies(R, P) for P, Q, R in models
    )
    entails_Q_implies_R = all(
        implies(Q, R) for P, Q, R in models
    )

    print("\nDoes KB entail R?", entails_R)
    print("Does KB entail R -> P?", entails_R_implies_P)
    print("Does KB entail Q -> R?", entails_Q_implies_R)


if __name__ == "__main__":
    main()
```
