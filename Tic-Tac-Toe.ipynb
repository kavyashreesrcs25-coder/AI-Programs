```python
import random
import ipywidgets as widgets
from IPython.display import display

board = [" "] * 9
game_over = False

def check_winner(player):
    wins = [
        (0,1,2), (3,4,5), (6,7,8),
        (0,3,6), (1,4,7), (2,5,8),
        (0,4,8), (2,4,6)
    ]

    for a, b, c in wins:
        if board[a] == board[b] == board[c] == player:
            return True

    return False


def show_board():
    print()
    print(board[0], "|", board[1], "|", board[2])
    print("--+---+---")
    print(board[3], "|", board[4], "|", board[5])
    print("--+---+---")
    print(board[6], "|", board[7], "|", board[8])
    print()


# Position selector
position = widgets.Dropdown(
    options=[
        ("Position 1", 1),
        ("Position 2", 2),
        ("Position 3", 3),
        ("Position 4", 4),
        ("Position 5", 5),
        ("Position 6", 6),
        ("Position 7", 7),
        ("Position 8", 8),
        ("Position 9", 9)
    ],
    description="Move:"
)

play_button = widgets.Button(
    description="PLAY",
    button_style="success"
)

output = widgets.Output()


def play_move(b):

    global game_over

    with output:

        if game_over:
            print("Game is already over.")
            return

        pos = position.value - 1

        if board[pos] != " ":
            print("Position already occupied!")
            return

        # Human move
        board[pos] = "X"

        print("\nYou selected position:", pos + 1)
        show_board()

        if check_winner("X"):
            print("🎉 HUMAN WINS!")
            game_over = True
            return

        if " " not in board:
            print("DRAW!")
            game_over = True
            return

        # Computer move
        empty = [i for i in range(9) if board[i] == " "]
        computer = random.choice(empty)

        board[computer] = "O"

        print("Computer selected position:", computer + 1)
        show_board()

        if check_winner("O"):
            print("🤖 COMPUTER WINS!")
            game_over = True
            return

        if " " not in board:
            print("DRAW!")
            game_over = True


play_button.on_click(play_move)

print("===== TIC-TAC-TOE =====")
print("You = X")
print("Computer = O")
print()
print("Select your position and click PLAY.")

show_board()

display(position, play_button, output)
```
