```python

def is_variable(x):
    return x[0].islower()

def main():
    print("UNIFICATION IN FIRST ORDER LOGIC")

    exp1 = input("Enter first expression: ")
    exp2 = input("Enter second expression: ")

    p1 = exp1.replace(")", "").split("(")
    p2 = exp2.replace(")", "").split("(")

    if p1[0].strip() != p2[0].strip():
        print("Unification Failed")
        return

    a1 = [x.strip() for x in p1[1].split(",")]
    a2 = [x.strip() for x in p2[1].split(",")]

    if len(a1) != len(a2):
        print("Unification Failed")
        return

    substitution = {}

    for x, y in zip(a1, a2):
        if x == y:
            continue

        if is_variable(x):
            if x in substitution and substitution[x] != y:
                print("Unification Failed")
                return
            substitution[x] = y

        elif is_variable(y):
            if y in substitution and substitution[y] != x:
                print("Unification Failed")
                return
            substitution[y] = x

        else:
            print("Unification Failed")
            return

    print("Substitution:", substitution)
    print("Unified Expression:", exp1.split("(")[0] +
          "(" + ", ".join(substitution.get(x, x) for x in a1) + ")")
    print("Unification Successful")


if __name__ == "__main__":
    main()
```
