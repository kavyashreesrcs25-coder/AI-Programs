```python
print("VACUUM CLEANER PROBLEM")

room_a = input("Enter Room A status (Clean/Dirty): ").capitalize()
room_b = input("Enter Room B status (Clean/Dirty): ").capitalize()

position = input("Enter vacuum position (A/B): ").upper()

print("\nInitial State")
print("Room A:", room_a)
print("Room B:", room_b)
print("Vacuum Position:", position)

# If vacuum is in Room A
if position == "A":

    if room_a == "Dirty":
        print("\nAction: SUCK")
        room_a = "Clean"
    else:
        print("\nRoom A is already clean")

    print("Action: MOVE RIGHT")
    position = "B"


# If vacuum is in Room B
if position == "B":

    if room_b == "Dirty":
        print("Action: SUCK")
        room_b = "Clean"
    else:
        print("Room B is already clean")


print("\nFinal State")
print("Room A:", room_a)
print("Room B:", room_b)

if room_a == "Clean" and room_b == "Clean":
    print("Both rooms are clean!")
```
