```python
from itertools import combinations

def calculate_cost(state):
    cost = 0

    for i in range(4):
        for j in range(i + 1, 4):
            if abs(state[i] - state[j]) == abs(i - j):
                cost += 1

    return cost


def hill_climbing():
    state = (3, 1, 2, 0)

    print("Initial State:", state)
    print("Initial Cost:", calculate_cost(state))

    while True:
        current_cost = calculate_cost(state)
        best_state = state
        best_cost = current_cost

        for i, j in combinations(range(4), 2):
            new_state = list(state)
            new_state[i], new_state[j] = new_state[j], new_state[i]
            new_state = tuple(new_state)

            cost = calculate_cost(new_state)

            print("Neighbor:", new_state, "Cost:", cost)

            if cost < best_cost:
                best_state = new_state
                best_cost = cost

        if best_cost >= current_cost:
            print("No better neighbor found.")
            break

        state = best_state

        print("Selected State:", state)
        print("Selected Cost:", best_cost)

        if best_cost == 0:
            break

    print("Final State:", state)
    print("Final Cost:", calculate_cost(state))

    if calculate_cost(state) == 0:
        print("Goal Reached!")


if __name__ == "__main__":
    hill_climbing()
```
