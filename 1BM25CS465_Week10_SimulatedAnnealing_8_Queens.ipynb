```python

import random
import math


def calculate_cost(board):
    conflicts = 0

    for i in range(len(board)):
        for j in range(i + 1, len(board)):

            if abs(board[i] - board[j]) == abs(i - j):
                conflicts += 1

    return conflicts


def generate_neighbor(board):
    neighbor = board.copy()
    i, j = random.sample(range(len(board)), 2)
    neighbor[i], neighbor[j] = neighbor[j], neighbor[i]
    return neighbor


def print_board(board):
    n = len(board)

    for row in range(n):
        line = ""

        for col in range(n):
            if board[col] == row + 1:
                line += " Q "
            else:
                line += " . "

        print(line)


def simulated_annealing(board, temperature, cooling_rate, iterations):

    current = board.copy()
    current_cost = calculate_cost(current)

    best = current.copy()
    best_cost = current_cost

    while temperature > 0.001 and best_cost != 0:

        for _ in range(iterations):

            next_state = generate_neighbor(current)
            next_cost = calculate_cost(next_state)

            delta_e = current_cost - next_cost

            if delta_e > 0:
                current = next_state
                current_cost = next_cost

            else:
                probability = math.exp(delta_e / temperature)

                if random.random() < probability:
                    current = next_state
                    current_cost = next_cost

            if current_cost < best_cost:
                best = current.copy()
                best_cost = current_cost

            if best_cost == 0:
                return best, best_cost

        temperature = temperature * cooling_rate

    return best, best_cost


def main():

    print("SIMULATED ANNEALING - N QUEENS")
    print("--------------------------------")

    n = int(input("Enter number of queens: "))

    if n < 4:
        print("Solution is not possible for N =", n)
        return

    print("\n1. Enter initial board")
    print("2. Generate random initial board")

    choice = int(input("Enter choice: "))

    if choice == 1:

        print("Enter", n, "numbers (permutation of 1 to", n, "):")
        initial = list(map(int, input().split()))

        if len(initial) != n or sorted(initial) != list(range(1, n + 1)):
            print("Invalid board!")
            return

    elif choice == 2:
        initial = list(range(1, n + 1))
        random.shuffle(initial)

    else:
        print("Invalid choice!")
        return

    temperature = float(input("Enter initial temperature: "))
    cooling_rate = float(input("Enter cooling rate (0 to 1): "))
    iterations = int(input("Enter iterations per temperature: "))
    restart_limit = int(input("Enter maximum restarts: "))

    print("\nInitial Board:")
    print(initial)
    print_board(initial)
    print("Initial Cost:", calculate_cost(initial))

    best_board = initial.copy()
    best_cost = calculate_cost(initial)

    for restart in range(restart_limit + 1):

        if restart == 0:
            start = initial.copy()
        else:
            start = list(range(1, n + 1))
            random.shuffle(start)

        print("\nRestart", restart + 1)

        result, cost = simulated_annealing(
            start, temperature, cooling_rate, iterations
        )

        print("Result:", result)
        print("Cost:", cost)

        if cost < best_cost:
            best_board = result.copy()
            best_cost = cost

        if cost == 0:
            break

    print("\nFinal Board:")
    print(best_board)
    print_board(best_board)
    print("Final Cost:", best_cost)

    if best_cost == 0:
        print("Solution Found!")
    else:
        print("Solution not found within restart limit.")


if __name__ == "__main__":
    main()
```
