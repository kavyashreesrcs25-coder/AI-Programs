```python

def build_tree(branching, values, index, level=0, path="R"):

    if level == len(branching):
        return values[index[0]], index[0] + 1

    children = []

    for i in range(branching[level]):
        child, index[0] = build_tree(
            branching, values, index, level + 1,
            path + "." + str(i + 1)
        )
        children.append(child)

    return children, index[0]


def get_leaf_paths(tree, path="R"):

    if not isinstance(tree, list):
        return [path]

    paths = []

    for i, child in enumerate(tree):
        paths.extend(get_leaf_paths(child, path + "." + str(i + 1)))

    return paths


def alpha_beta(tree, alpha, beta, maximizing, path="R", pruned=None):

    if pruned is None:
        pruned = []

    if not isinstance(tree, list):
        print("Visited leaf", path, "=", tree)
        return tree

    if maximizing:
        value = float("-inf")

        for i, child in enumerate(tree):

            child_path = path + "." + str(i + 1)

            result = alpha_beta(
                child, alpha, beta, False, child_path, pruned
            )

            value = max(value, result)
            alpha = max(alpha, value)

            if alpha >= beta:

                for remaining in tree[i + 1:]:
                    start = i + 2
                    pruned.extend(
                        get_leaf_paths(remaining, path + "." + str(start))
                    )
                    start += 1

                print("Pruned at", path)
                break

        return value

    else:
        value = float("inf")

        for i, child in enumerate(tree):

            child_path = path + "." + str(i + 1)

            result = alpha_beta(
                child, alpha, beta, True, child_path, pruned
            )

            value = min(value, result)
            beta = min(beta, value)

            if beta <= alpha:

                for j in range(i + 1, len(tree)):
                    pruned.extend(
                        get_leaf_paths(tree[j], path + "." + str(j + 1))
                    )

                print("Pruned at", path)
                break

        return value


def main():

    print("ALPHA-BETA PRUNING")
    print("--------------------")

    levels = int(input("Enter number of internal levels: "))

    branching = []

    for i in range(levels):
        b = int(input("Enter branching factor for level " + str(i + 1) + ": "))
        branching.append(b)

    total_leaves = 1

    for b in branching:
        total_leaves *= b

    print("\nTotal leaf nodes:", total_leaves)

    values = []

    for i in range(total_leaves):
        value = int(input("Enter leaf value " + str(i + 1) + ": "))
        values.append(value)

    index = [0]

    tree, _ = build_tree(branching, values, index)

    print("\nStarting Alpha-Beta Search")
    print("-------------------------")

    pruned = []

    result = alpha_beta(
        tree, float("-inf"), float("inf"), True, "R", pruned
    )

    print("\nOptimal Root Value:", result)

    if pruned:
        print("Pruned leaf paths:", pruned)
    else:
        print("No leaf nodes pruned.")

    print("Search completed.")


if __name__ == "__main__":
    main()
```
